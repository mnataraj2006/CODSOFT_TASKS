## 1. Objective
The objective of this project is to inspect, clean, validate, and preprocess the retail sales dataset using Python and Pandas. The goals are:
- Load and inspect the dataset's structural layout.
- Check for missing values and handle them appropriately if found.
- Detect and remove duplicate records.
- Investigate categorical and text fields for casing, whitespaces, or inconsistencies.
- Correct any inappropriate data types.
- Validate numerical values and mathematical relations between columns (e.g., verifying `revenue` consistency).
- Export the final, cleaned dataset to `Data/cleaned_data.csv` for downstream analytics.

## 2. Import Libraries
Import the necessary python libraries required for this project. We'll use Pandas for data manipulation and analysis, and NumPy for numerical operations if needed.

```python
import pandas as pd
import numpy as np
```

## 3. Load Dataset
Load the raw retail sales dataset from the `Data` directory and display the first few rows to inspect its layout.

```python
# Load raw dataset
df = pd.read_csv('Data/Raw_data_RetailSales.csv')

# Display the first few rows
df.head()
```

## 4. Dataset Overview
We will inspect the basic structural properties of the dataset including dimensions, column names, data types, and check both numerical and categorical summary statistics.

```python
# Shape of dataset
print("Dataset Dimensions (Rows, Columns):", df.shape)

# Columns and types
print("\nDataset Info:")
df.info()

# Basic descriptive statistics for numerical columns
print("\nNumerical Summary:")
display(df.describe())

# Descriptive statistics for categorical columns
print("\nCategorical Summary:")
display(df.describe(include=['object']))

# Check unique value counts for each column
print("\nUnique Values per Column:")
display(df.nunique())
```

## 5. Missing Value Analysis
We calculate the count and percentage of missing values in each column to identify potential gaps in the raw data.

```python
# Missing value counts and percentages
missing_counts = df.isnull().sum()
missing_pct = df.isnull().mean() * 100

missing_summary = pd.DataFrame({
    'Column': df.columns,
    'Missing Count': missing_counts.values,
    'Missing Percentage (%)': missing_pct.values
}).set_index('Column')

print("Missing Values Summary:")
display(missing_summary)
```

## 6. Duplicate Analysis
We check for exact duplicate rows in the dataset. If any duplicate rows exist, we will report them and remove them.

```python
# Count duplicates
duplicate_count = df.duplicated().sum()
print(f"Original row count: {len(df)}")
print(f"Total duplicates found: {duplicate_count}")

if duplicate_count > 0:
    # Display sample duplicate rows
    print("\nSample duplicate records:")
    display(df[df.duplicated(keep=False)].head(10))
    
    # Remove duplicate records
    df_cleaned = df.drop_duplicates().copy()
    print(f"Duplicates removed: {duplicate_count}")
    print(f"Row count after duplicate removal: {len(df_cleaned)}")
else:
    df_cleaned = df.copy()
    print("No duplicate records were identified.")
```

## 7. Data Consistency Analysis
We inspect the categorical columns `product_category`, `region`, and `payment_method` for any common formatting issues, such as leading/trailing whitespaces or inconsistent casing (e.g. Card vs card).

```python
# Casing and whitespace audit for categorical columns
categorical_cols = ['product_category', 'region', 'payment_method']
for col in categorical_cols:
    unique_vals = df_cleaned[col].unique()
    print(f"\nUnique values in '{col}': {list(unique_vals)}")
    
    # Check if stripping whitespaces changes unique values count
    stripped = [str(x).strip() for x in unique_vals]
    if len(unique_vals) != len(set(stripped)):
        print(f"  -> INFO: Found leading or trailing whitespaces in '{col}'. Correcting them.")
        df_cleaned[col] = df_cleaned[col].str.strip()
        
    # Check if lowercase conversion changes unique values count
    lowercased = [str(x).lower() for x in unique_vals]
    if len(unique_vals) != len(set(lowercased)):
        print(f"  -> INFO: Found inconsistent capitalization in '{col}'. Correcting them.")
        # Capitalize values properly to maintain readability
        df_cleaned[col] = df_cleaned[col].str.title()

print("\nData consistency audit completed.")
```

## 8. Missing Value Treatment
If any missing values had been identified in Step 5, we would treat them using strategies like mean/median imputation, mode imputation, or deletion. Since 0 missing values were found, no missing value treatment is necessary.

```python
# Documenting missing values status
print("No missing values were detected in the dataset. Missing value treatment is not required.")
```

## 9. Data Type Correction
The `order_date` column is currently stored as a string (object). We need to parse and convert it to a proper datetime format (`datetime64[ns]`) for time-based analytics.

```python
# Verify and convert data types
print("Original order_date type:", df_cleaned['order_date'].dtype)

# Convert to datetime using explicit format
df_cleaned['order_date'] = pd.to_datetime(df_cleaned['order_date'], format='%m/%d/%Y', errors='coerce')

# Verify the changes
print("Updated order_date type:", df_cleaned['order_date'].dtype)
print("Coerced null/NaT count:", df_cleaned['order_date'].isnull().sum())
```

## 10. Data Validation
We validate that the numeric columns have logical ranges (e.g. positive prices, quantities, valid ratings) and check if the `revenue` matches the formula `quantity * unit_price * (1 - discount)`.

```python
# Validate numeric boundaries
validation_checks = {
    'positive_quantity': (df_cleaned['quantity'] >= 1).all(),
    'positive_unit_price': (df_cleaned['unit_price'] > 0).all(),
    'valid_discount': ((df_cleaned['discount'] >= 0.0) & (df_cleaned['discount'] <= 1.0)).all(),
    'valid_delivery_days': (df_cleaned['delivery_days'] >= 0).all(),
    'valid_customer_rating': ((df_cleaned['customer_rating'] >= 1.0) & (df_cleaned['customer_rating'] <= 5.0)).all()
}

for name, passed in validation_checks.items():
    print(f"Validation '{name}': {'PASSED' if passed else 'FAILED'}")

# Verify mathematical relation: revenue = quantity * unit_price * (1 - discount)
calculated_rev = df_cleaned['quantity'] * df_cleaned['unit_price'] * (1 - df_cleaned['discount'])
revenue_differences = (df_cleaned['revenue'] - calculated_rev).abs()

print(f"\nMax difference between actual and formula revenue: {revenue_differences.max()}")
print(f"Mean difference: {revenue_differences.mean()}")

# If the differences are negligible (due to float rounding), the calculation is correct.
if revenue_differences.max() < 0.01:
    print("Verification: The revenue column is mathematically consistent with the subtotal formula.")
else:
    print("Warning: Mathematical inconsistency found in the revenue column.")
```

## 11. Final Data Quality Check
Let's print the final dataset structure, verify the data types, checking for any remaining nulls or duplicates, and output a concise Before/After cleaning summary.

```python
print("Final Dataset Shape:", df_cleaned.shape)
print("Final Missing Values count per column:")
print(df_cleaned.isnull().sum())
print("Final Duplicate Rows count:", df_cleaned.duplicated().sum())

# Create a before vs after summary table
summary_data = {
    'Metric': ['Rows', 'Columns', 'Missing Values', 'Duplicate Rows'],
    'Before': [df.shape[0], df.shape[1], df.isnull().sum().sum(), df.duplicated().sum()],
    'After': [df_cleaned.shape[0], df_cleaned.shape[1], df_cleaned.isnull().sum().sum(), df_cleaned.duplicated().sum()]
}
df_summary = pd.DataFrame(summary_data)
display(df_summary)

# Display sample of preprocessed data
df_cleaned.head()
```

## 12. Export Cleaned Dataset
Export the preprocessed and verified dataset to `Data/cleaned_data.csv` using `index=False` to avoid saving the DataFrame row index.

```python
# Export cleaned dataset
df_cleaned.to_csv('Data/cleaned_data.csv', index=False)
print("Cleaned data successfully saved to 'Data/cleaned_data.csv'")
```
